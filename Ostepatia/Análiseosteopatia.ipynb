# 1. Efficacy and safety of osteopathic manipulative treatment (OMT): overview of systematic reviews

# 🎯 Objetivo

Reunir revisões sistemáticas (SRs) e meta-análises (MAs) de ensaios clínicos randomizados (RCTs) para avaliar a eficácia e a segurança do tratamento manipulado osteopático (OMT), até novembro de 2021.

# 📊 Dados e Resultados

Foram incluídas 9 SRs/MAs publicadas entre 2013 e 2020, abrangendo 55 estudos primários com um total de 3.740 participantes.

As condições analisadas incluíram: lombalgia aguda/crônica não específica, cervicalgia, dor crônica não‑câncer, cefaleia primária, pediátrica e síndrome do intestino irritável (IBS).

Eficácia clínica moderada para OMT em: dor e função — especialmente em lombalgia e cervicalgia não específicas, e em dor crônica não relacionada ao câncer. Para cefaleia e IBS, os resultados foram inconsistentes e baseados em evidência limitada ou conflitante.

A maioria dos estudos teve qualidade metodológica baixa ou criticamente baixa segundo a ferramenta AMSTAR‑2. Os RCTs apresentaram alto risco de viés e quase não registraram eventos adversos (quando registrados, foram leves).

# ✔️ Conclusões

Evidência promissora para eficácia do OMT em certas condições musculoesqueléticas.

Para outras condições (pediátricas, cefaleias, IBS), os resultados permanecem inconclusivos.

Necessidade de ensaios clínicos bem desenhados com amostras maiores, cegamento rigoroso e relatos transparentes de eventos adversos.

# 2. The effectiveness of visceral osteopathy in pain, disability, and physical function in patients with low‑back pain: systematic review and meta‑analysis

# 🎯 Objetivo

Avaliar se a osteopatia visceral melhora dor, incapacidade funcional ou desempenho físico em pacientes com lombalgia.

# 📊 Dados e Resultados

Baseada em 5 estudos randomizados, com 268 pacientes ao todo.

A qualidade metodológica variou de alta a baixa, mas o risco de viés foi alto em todos os estudos incluídos.

Dor: SMD = –0.53 (IC 95%: –1.09 a 0.03), com alta heterogeneidade (I² = 78 %).

Incapacidade funcional: SMD = –0.08 (IC 95%: –0.44 a 0.27; I² = 0 %).

Função física: SMD = –0.26 (IC 95%: –0.62 a 0.10; I² = 0 %).

# ✔️ Conclusão

Nenhum benefício estatisticamente significativo foi observado em dor, incapacidade ou função física.

A combinação de tamanhos amostrais pequenos e alto risco de viés limita fortemente a confiabilidade dos achados.

Conclui-se que não há evidências robustas para apoiar a eficácia da osteopatia visceral no tratamento da lombalgia.

# 🧾 Comparação entre os dois estudos

| Técnica                     | Eficácia Clínica                                                       | Qualidade metodológica                                                                            | Considerações finais                                                 |
| --------------------------- | ---------------------------------------------------------------------- | ------------------------------------------------------------------------------------------------- | -------------------------------------------------------------------- |
| **OMT (manipulação geral)** | Eficácia moderada em dor e função nos casos de lombalgia e cervicalgia | SRs/MAs amplas porém de baixa qualidade; risco de viés alto e relato limitado de efeitos adversos | Promissora para uso musculoesquelético; exige estudos mais rigorosos |
| **Visceral Osteopathy**     | **Sem eficácia comprovada** em dor ou incapacidade                     | Estudos reduzidos, amostras pequenas, alto viés                                                   | Não há suporte científico confiável para prescrição                  |

# 🔍 Conclusão Final

O OMT geral apresenta evidência moderada de eficácia para dor musculoesquelética (lombalgia/cruralgia e cervicalgia), com melhora funcional.

A osteopatia visceral não demonstrou benefícios claros em condições de lombalgia, e a evidência existente é fraca e metodologicamente limitada.

O nível de certeza dos dois tipos de intervenção é comprometedido por falhas metodológicas: pequena amostra, alto risco de viés, falta de cegamento e escassez de relatos sistemáticos de efeitos adversos.

# 1. Osteopatia Craniada (Cranial Osteopathy) 🧠

# 🎯 Objetivo

Avaliar a confiabilidade dos diagnósticos e a eficácia clínica das técnicas da osteopatia craniana.

# 🧪 Metodologia

Buscas até junho de 2016 em MEDLINE, PEDro, OSTMED.DR, Cochrane, Google Scholar, entre outros.

Para confiabilidade: estudos com medição intra e interexaminadores; para eficácia: apenas RCTs ou ensaios crossover.

Excluídos estudos sem texto completo, não humanos, ou com técnicas não claramente cranianas.

# 📌 Resultados

Confiabilidade diagnóstica: nenhum dos estudos mostrou reprodutibilidade confiável (ICC < 0.75 e κ < 0.81 em geral). 

Alto risco de viés (falta de cegamento, variabilidade no treinamento).

Eficácia clínica: apenas 3 estudos com baixo risco de viés – e mesmo nesses, não foi considerado efeito não-específico do tratamento.

A maioria dos RCTs demonstrou alto ou incerto risco de viés.

# ✔️ Conclusão

Evidência de confiabilidade e eficácia é praticamente inexistente. Estudos existentes são metodologicamente fragilizados e insuficientes para respaldar o uso clínico da osteopatia cranial.

# 2. Osteopatia Visceral (Visceral Osteopathy)

# 🎯 Objetivo

Examinar a confiabilidade dos diagnósticos e a eficácia clínica das técnicas viscerais em condições diversas (incluindo lombalgia).

# 🧪 Metodologia

Buscas até dezembro de 2017 em várias bases (MEDLINE, Cochrane, etc.).

Critérios similares: confiabilidade por inter/intraexaminador; eficácia apenas RCTs ou crossover em pacientes com condições clínicas.

# 📌 Resultados

Confiabilidade diagnóstica: todas as técnicas demonstraram confiabilidade inadequada, com alto risco de viés metodológico (estatística inadequada, ausência de cegamento).

Eficácia clínica: o estudo com menor risco de viés não mostrou diferença significativa entre tratamento visceral + fisioterapia versus fisioterapia isolada. Estudos restantes foram de qualidade baixa ou veementemente criticados.

# ✔️ Conclusão

A evidência sobre confiabilidade e eficácia da osteopatia visceral é insuficiente para justificar sua utilização clínica, especialmente no manejo da lombalgia.

# 3. Osteopatia Manipulativa Geral (OMT)

# 🎯 Objetivo

Compilar evidências de revisões sistemáticas e meta-análises sobre OMT relacionadas a dor musculoesquelética e outros contextos clínicos até 2021.

# 📌 Resultados Relevantes

Evidência moderada para redução de dor e melhora funcional em lombalgia aguda/crônica não específica (inclusive pós-parto) e cervicalgia. A qualidade geral dos estudos era baixa ou criticamente baixa segundo AMSTAR‑2.

Para condições como IBS, cefaleia primária e contexto pediátrico, os resultados foram inconsistentes ou baseados em amostras pequenas e heterogeneidade elevada.

# ✔️ Conclusão

A OMT mostra-se promissora para dor musculoesquelética, mas exige ensaios clínicos com desenhos metodológicos robustos, maior tamanho amostral, cegamento adequado e relatos padronizados de segurança.

# ✅ Considerações Finais

A OMT tradicional apresenta alguma base para uso em dor musculoesquelética, com benefícios moderados já observados em meta-análises, mas ainda faltam estudos mais confiáveis.

As técnicas viscerais e cranianas não têm respaldo científico sólido, nem para diagnóstico nem para tratamento.

Há fortes recomendações para que futuras pesquisas usem protocolos padronizados (TIDieR/CONSORT), cegamento rigoroso, amostras adequadas e documentação completa de efeitos adversos.

```python
import matplotlib.pyplot as plt
import numpy as np

# Dados mais detalhados das análises citadas
dados = {
    "OMT geral (lombalgia/cervicalgia)": {"Eficácia": 2, "Qualidade dos estudos": 1, "Segurança": 2},
    "OMT geral (pediatria/IBS/cefaleia)": {"Eficácia": 1, "Qualidade dos estudos": 1, "Segurança": 2},
    "Osteopatia visceral": {"Eficácia": 0, "Qualidade dos estudos": 0, "Segurança": 1},
    "Osteopatia craniana": {"Eficácia": 0, "Qualidade dos estudos": 0, "Segurança": 1}
}

categorias = list(dados.keys())
metricas = list(next(iter(dados.values())).keys())

fig, ax = plt.subplots(figsize=(10, 6))
bar_width = 0.25
x = np.arange(len(categorias))

cores = ["#4CAF50", "#2196F3", "#FF9800"]

for i, metrica in enumerate(metricas):
    valores = [dados[cat][metrica] for cat in categorias]
    barras = ax.bar(x + i * bar_width, valores, width=bar_width, label=metrica, alpha=0.9, color=cores[i])
    
    # Adiciona rótulos de valores
    for barra in barras:
        altura = barra.get_height()
        ax.annotate(f'{altura}',
                    xy=(barra.get_x() + barra.get_width() / 2, altura),
                    xytext=(0, 3),
                    textcoords="offset points",
                    ha='center', va='bottom', fontsize=9, fontweight="bold")

# Configurações do gráfico
ax.set_xlabel("Técnicas / Condições analisadas", fontsize=12, fontweight="bold")
ax.set_ylabel("Nível (0=nenhuma, 3=alta)", fontsize=12, fontweight="bold")
ax.set_title("Resumo das análises científicas sobre osteopatia", fontsize=14, fontweight="bold")
ax.set_xticks(x + bar_width)
ax.set_xticklabels(categorias, rotation=25, ha="right")
ax.set_ylim(0, 3.5)
ax.legend()

plt.tight_layout()
plt.show()
```

OMT geral para lombalgia e cervicalgia é a técnica com melhor desempenho geral:

Eficácia moderada (nível 2) para redução de dor e melhora funcional.

Segurança considerada boa (nível 2).

Qualidade metodológica baixa (nível 1), ou seja, a evidência existe, mas é fragilizada por limitações nos estudos.

OMT geral para pediatria, IBS e cefaleia apresenta eficácia baixa/inconclusiva (nível 1) e mesma segurança e qualidade metodológica do grupo anterior.

Osteopatia visceral e osteopatia craniana têm nenhuma eficácia comprovada (nível 0) e qualidade metodológica muito baixa (nível 0), embora sejam geralmente seguras (nível 1), com raros efeitos adversos relatados.

A única aplicação da osteopatia com evidência moderada e consistente é o OMT geral para dor musculoesquelética (especialmente lombalgia e cervicalgia). Já as técnicas viscerais e cranianas não têm suporte científico robusto para eficácia, e a qualidade dos estudos existentes é insuficiente para sustentar recomendações clínicas sólidas.

# 💡 Considerações Gerais:

**1. OMT geral (manipulação osteopática tradicional)**

Onde há evidência moderada: dor musculoesquelética, especialmente lombalgia e cervicalgia.

Benefícios observados: melhora funcional e redução de dor.

**Limitação:** a qualidade metodológica dos estudos é baixa (falta de cegamento, amostras pequenas, risco de viés).

**Segurança:** geralmente segura, com efeitos adversos leves e raros.

**2. OMT em pediatria, cefaleia e síndrome do intestino irritável**

Evidência fraca ou inconclusiva: estudos mostram resultados inconsistentes.

Possível efeito placebo não pode ser descartado.

**Necessidade:** pesquisas maiores e mais rigorosas.

**3. Osteopatia visceral**

Nenhuma eficácia comprovada em dor, incapacidade ou função física.

**Qualidade dos estudos:** muito baixa, com alto risco de viés.

**Uso clínico:** não sustentado por evidência robusta.

**4. Osteopatia craniana**

Confiabilidade diagnóstica muito baixa (avaliações não reprodutíveis entre examinadores).

Eficácia clínica não demonstrada.

Frequentemente classificada na literatura como prática sem base científica sólida.
