# Revisão da Literatura:

**Câncer de Mama**

É a segunda neoplasia maligna mais diagnosticada no mundo e a que mais mata mulheres no Brasil.

Diagnóstico precoce é determinante para o prognóstico.

Tratamentos variam conforme estadiamento e incluem cirurgia, quimio, rádio e hormonioterapia.

Complicações pós-mastectomia: dor, linfedema, fraqueza muscular, restrição de movimento, alterações posturais.

**Fisioterapia no Pós-operatório:**

Deve ser precoce e individualizada, visando:

Recuperar amplitude de movimento.

Reduzir complicações pulmonares.

Reintegrar paciente às atividades cotidianas.

Técnicas: cinesioterapia, drenagem linfática manual, exercícios respiratórios, incentivadores respiratórios, mobilização cicatricial.

Destaca-se o papel na prevenção e manejo de disfunções cardiorrespiratórias decorrentes da cirurgia.
