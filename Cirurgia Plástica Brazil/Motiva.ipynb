| **Recurso**                | **Detalhe Técnico**                                                                                                         | **Benefício para a Paciente**                                                                                                                                         |

# SmoothSilk® / SilkSurface® (A Superfície)

**Este é um dos diferenciais mais importantes da Motiva, focando na segurança e biocompatibilidade.**

**Nano-superfície Avançada:** Ao contrário dos implantes texturizados (que tiveram problemas de segurança histórica) e dos implantes lisos tradicionais, a Motiva criou uma superfície avançada que é lisa (avançadamente nano-texturizada) em 360°.

**Biocompatibilidade:** É cientificamente projetada para promover a biointegração e minimizar a reação inflamatória do corpo.

**Redução de Complicações:** Esta superfície tem demonstrado em estudos um perfil favorável, com taxas muito baixas de contratura capsular (endurecimento da cápsula ao redor do implante) e está associada a um risco significativamente reduzido do **Linfoma Anaplásico de Células Gigantes (BIA-ALCL),** uma preocupação de segurança global.

# TrueMonobloc® e TrueMonobloc+® (O Envoltório)

Refere-se à arquitetura do envoltório (shell) e sua conexão com o gel e o patch traseiro.

**Estrutura Coesa:** É uma tecnologia proprietária que une o envoltório do implante, o gel e a base em uma estrutura única e coesa.

**Maior Resistência e Flexibilidade:** Essa união confere ao implante alta resistência à ruptura, ao mesmo tempo que mantém a flexibilidade.

**MinimalScar®:** A flexibilidade aprimorada (especialmente no **TrueMonobloc+® no Ergonomix2®**) permite que o implante seja comprimido em um grau mais alto, facilitando a inserção através de incisões cirúrgicas menores, o que apoia a **técnica Motiva MinimalScar®.**

# BluSeal® / BluSeal+® (Barreira de Segurança)

É um indicador visual de segurança no envoltório do implante.

**Camada de Barreira:** É uma camada protetora levemente tingida de azul com corantes biocompatíveis.

**Função:** Atua como uma barreira que minimiza e impede a difusão ("sangramento") do gel de silicone para os tecidos circundantes, um problema comum em implantes mais antigos.

**Controle de Qualidade:** O tom azulado permite que o cirurgião inspecione visualmente a integridade e uniformidade dessa camada antes da implantação, garantindo a presença da barreira de segurança.

# Qid® (Microchip de Identificação)

**O Qid®** é um recurso de rastreabilidade exclusivo e inovador.

**Tecnologia RFID (Radio Frequency Identification):** É um microtransponder passivo e sem bateria, pequeno e seguro, incorporado na parte traseira do implante.

**Rastreabilidade:** Armazena o Número de Série Eletrônico (ESN) e informações sobre o dispositivo.

**Verificação Pós-Cirurgia:** Permite que o cirurgião use um leitor portátil externo para confirmar os detalhes do implante Motiva após a cirurgia (mesmo sem o cartão de identificação impresso da paciente), garantindo a rastreabilidade e a ativação da garantia estendida.

**A Motiva investe pesadamente na ciência da superfície (SmoothSilk®) para a segurança e na reologia do gel (ProgressiveGel Ultima®) para a naturalidade e sensação tátil, posicionando-se como uma opção de alta performance e segurança no mercado global.**

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Dados de Receita (em milhões de USD)
# Estimativa de Receita 2024 (Aproximada)
receita_2024_est = 158.0
# Projeção de Receita 2025 (Exceder $210M)
receita_2025_proj = 215.0
# Projeção de Receita 2026 (Consenso de Analistas)
receita_2026_proj = 261.0

anos = ['2024 (Est.)', '2025 (Proj.)', '2026 (Proj.)']
receita = [receita_2024_est, receita_2025_proj, receita_2026_proj]

# GRÁFICO 1: Crescimento da Receita
plt.figure(figsize=(8, 5))
barras = plt.bar(anos, receita, color=['#FFA07A', '#FF4500', '#B22222'])

# Adicionar rótulos de dados
for barra in barras:
    yval = barra.get_height()
    plt.text(barra.get_x() + barra.get_width()/2.0, yval + 3, f'${yval:.1f}M', ha='center', va='bottom', fontsize=10)

plt.title('Projeção de Crescimento da Receita Anual (Establishment Labs)', fontsize=14, fontweight='bold')
plt.ylabel('Receita (Milhões de USD)', fontsize=12)
plt.xlabel('Ano Fiscal', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.yticks(np.arange(0, 300, 50))
plt.show()


# Dados de Margem Bruta (Gross Margin) e EBITDA
# Dados baseados em performance real e metas
data = {
    'Trimestre': ['T2-2024', 'T3-2024', 'T2-2025', 'T3-2025', 'T4-2025 (Meta)', '2026 (Meta)'],
    'Margem Bruta (%)': [66.5, 68.0, 69.2, 70.1, 70.5, 71.0],
    'EBITDA (Milhões USD)': [-4.5, -2.0, -0.5, 1.2, 2.0, 5.0]
}
df = pd.DataFrame(data)

# GRÁFICO 2: Margem Bruta e EBITDA
fig, ax1 = plt.subplots(figsize=(10, 6))

# Eixo Y1 (Margem Bruta)
ax1.plot(df['Trimestre'], df['Margem Bruta (%)'], marker='o', color='forestgreen', label='Margem Bruta (%)', linestyle='-', linewidth=2)
ax1.set_xlabel('Período de Referência')
ax1.set_ylabel('Margem Bruta (%)', color='forestgreen', fontsize=12)
ax1.tick_params(axis='y', labelcolor='forestgreen')
ax1.grid(axis='y', linestyle='--', alpha=0.5)

# Adicionar rótulos de dados (Margem Bruta)
for i, txt in enumerate(df['Margem Bruta (%)']):
    ax1.annotate(f'{txt}%', (df['Trimestre'][i], df['Margem Bruta (%)'][i]), textcoords="offset points", xytext=(0,10), ha='center', color='forestgreen')

# Eixo Y2 (EBITDA)
ax2 = ax1.twinx()
ax2.bar(df['Trimestre'], df['EBITDA (Milhões USD)'], color='teal', alpha=0.4, label='EBITDA (Milhões USD)')
ax2.set_ylabel('EBITDA (Milhões USD)', color='teal', fontsize=12)
ax2.tick_params(axis='y', labelcolor='teal')

# Adicionar rótulos de dados (EBITDA)
for i, val in enumerate(df['EBITDA (Milhões USD)']):
    ax2.annotate(f'${val:.1f}M', (df['Trimestre'][i], val), textcoords="offset points", xytext=(0, -15 if val < 0 else 5), ha='center', color='black' if val < 0 else 'teal')

ax2.axhline(0, color='gray', linestyle='-') # Linha de EBITDA zero

plt.title('Evolução da Margem Bruta e EBITDA (Establishment Labs) - Consolidando a Rentabilidade', fontsize=14, fontweight='bold')
fig.tight_layout()
plt.show()
```
