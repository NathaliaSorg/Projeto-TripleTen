# Estudo pós-comercialização de implantes mamários revestidos com espuma de poliuretano e gel de silicone da Silimed: resultados provisórios (2018-2024)

# Resumo
**Contexto:** Os implantes mamários revestidos com espuma de poliuretano (PU) de gel de silicone Silimed são utilizados em todo o mundo. Há ampla experiência clínica com o uso prolongado desses implantes, mas as evidências científicas sobre o tema ainda são limitadas. Este relatório avaliou os dados provisórios sobre o desempenho e a segurança desses implantes.

**Métodos:** Este é um ensaio clínico pós-comercialização, fase IV, aberto, não randomizado e de braço único. Os participantes foram selecionados consecutivamente com base em sua frequência no centro de pesquisa durante o período de inscrição. Os principais critérios de elegibilidade foram aumento primário ou de revisão e a ausência de condições que pudessem aumentar o risco de eventos adversos em curto prazo. Os implantes mamários PU da Silimed foram a intervenção. A segurança foi avaliada pelo monitoramento de eventos adversos. Os desfechos de maior interesse foram ruptura do implante, reoperação e contratura capsular. A eficácia foi medida pela satisfação, e a qualidade de vida foi medida pelo questionário BEQ55. Dados demográficos do paciente, detalhes cirúrgicos, características do implante e resultados pós-operatórios foram registrados.

**Resultados:** Este estudo incluiu 342 pacientes. A taxa de satisfação foi de 88,69% para o aumento primário e 86,00% para a revisão. Após 5 anos, não foram relatados casos de contratura capsular ou ruptura do implante. Reoperações ocorreram em 1,76%: 5 envolveram substituição do implante e 1 não. Os seromas identificados durante o estudo ocorreram dentro de 8 semanas após a cirurgia; portanto, não houve casos de linfoma anaplásico de grandes células associado a implantes mamários.

**Conclusões:** Os implantes mamários de PU Silimed demonstraram segurança favorável e alta eficácia ao longo do período de acompanhamento. A baixa incidência de contratura capsular e a ausência de linfoma anaplásico de grandes células associado ao implante mamário corroboram o uso de implantes de PU no aumento mamário.

# Pontos Fortes

**1.** Protocolo registrado e aprovação ética; reduz algumas fontes de viés típicas de estudos puramente retrospectivos.

**2.** Uso de instrumentos validados de resultado relatado pelo paciente (BEQ55, RSES) além de dados clínicos.

# Limitações (por que comprometem a força da evidência):

**1.** Desenho sem grupo de controlo; impossibilita comparar a performance das próteses Silimed com outras marcas/superfícies; não há randomização. Isso limita muito inferências causais.

**2.** Seguimento de longo prazo reduzido; apenas 5 pacientes com 5 anos completos; portanto a afirmação “após 5 anos, nenhum caso…” refere-se a um número muito pequeno e não permite detectar eventos raros ou de aparecimento tardio (p.ex. BIA-ALCL costuma aparecer entre 3–14 anos; mediana 8 anos).

**3.** Detecção incompleta de rupturas assintomáticas; apenas 1/3 da amostra foi selecionada para ressonância magnética seriada; como muitas rupturas são assintomáticas, a vigilância limitada pode subestimar rupturas reais.

**4.** Patrocínio integral pelo fabricante + autores com vínculo; risco de viés de reporte, seleção e análise (o próprio artigo declara o patrocínio da Silimed). Estudos financiados pela indústria tendem a mostrar resultados mais favoráveis; é um sinal para interpretar com cuidado.

**5.** Single-center (clínica em RJ) e seleção dos participantes; enrolamento consecutivo de pacientes que optaram pelo implante Silimed; isso reduz generalizabilidade e aumenta chance de viés por seleção.

**6.** Dados/brutos não públicos; o artigo indica que os datasets gerados/analisados não estão publicamente disponíveis, o que dificulta verificação independente e reanálise.

# O que a literatura independente diz:

Revisões sistemáticas e séries longas históricas sugerem que próteses poliuretano-recobertas frequentemente apresentam menor incidência de capsular contracture em comparação com implantes texturizados/suaves em vários estudos, mas a qualidade e heterogeneidade dos estudos variam e benefícios podem diminuir com o tempo.

A associação entre superfícies texturizadas (incluindo algumas poliuretano) e BIA-ALCL é complexa e ainda estudada; episódios são raros, frequentemente com latência de vários anos, por isso são necessários seguimentos longos e amplos para estimar riscos reais.

# Confiabilidade:

Confiabilidade para segurança/eficácia de longo prazo (≥5–10 anos) e para eventos raros (ruptura silenciosa, BIA-ALCL, complicações tardias): (NÃO); o estudo não fornece evidência robusta para essas conclusões por causa do baixo número de pacientes com seguimento de 4–5 anos, vigilância por imagem limitada e financiamento do fabricante. Portanto, as frases do artigo sobre “ausência de rupturas ou BIA-ALCL após 5 anos” devem ser interpretadas com extrema cautela e como resultados preliminares/interim, não como prova definitiva.

# Classificação geral do risco de viés (ROBINS-I):

**Confounding:** Serious; sem grupo de controlo, não ajusta fatores clínicos relevantes.

**Seleção de participantes:** Serious — single-center, critérios de inclusão favorecem pacientes de menor risco.

**Classificação da intervenção:** Low/Moderate; implante bem definido, mas técnicas cirúrgicas variam.

**Desvios da intervenção:** Moderate; open-label, patrocinado pelo fabricante, possível influência em condutas.

**Dados ausentes:** Serious; seguimento completo só até 1 ano para todos; apenas poucos pacientes com 5 anos.

**Medição dos desfechos:** Moderate; rupturas assintomáticas podem ter sido subdetectadas; BIA-ALCL exige latência >7 anos.

**Seleção de resultados relatados:** Serious; estudo interim, financiado 100% pela Silimed, dados brutos não públicos.

# Conclusão:

Evidência razoável apenas para resultados de curto prazo (≤1 ano).

Não confiável para segurança/eficácia de longo prazo ou eventos raros (ruptura silenciosa, BIA-ALCL).

Patrocínio integral da Silimed e perdas de seguimento reforçam risco de viés.

**Segurança a longo prazo/eventos raros:** não confiável para excluir riscos tardios como BIA-ALCL ou rupturas silenciosas a latência média do BIA-ALCL é de 7–10 anos, portanto é necessário seguimento muito mais longo e amostras maiores para estimar risco.
